```python
def binarySearchSqrt(target):
  l = 0; r = target
  while l <= r :
    middle = (l+r) // 2
    if middle**2 > target:
      r = middle- 1
      continue
    if middle**2 < target:
      l = middle+ 1
      continue
    return middle
  return r


def copyTime(n, x, y):
  l = 0
  r = (n - 1) * max(x, y)
  while l + 1 < r:
    mid = (r + l) // 2
    if mid//x + mid//y < n-1:
      l = mid
    else:
      r = mid
  return r + min(x, y)


def feedAnimals(animals, food):
  if len(animals) == 0 or len(food) == 0:
    return 0
  animals.sort()
  food.sort()
  count = 0
  for f in food:
    if f >= animals[count]:
      count += 1
    if count == len(animals):
      break
  return count



def extraLetter(a: str, b: str) -> str:
    hashMapA = {}

    # O(n)
    for i in range(len(a)):
        char = a[i]
        hashMapA[char] = hashMapA.get(char, 0) + 1

    # O(n)
    for i in range(len(b)):
        char = b[i]
        if char in hashMapA:
            hashMapA[char] -= 1
            # как только счетчик становится равен 0 - удаляем элемент из хеш-таблицы
            if hashMapA[char] == 0:
                del hashMapA[char]
            continue

        # если мы удалили ключ со значением char или же этой буквы никогда не было в таблице
        return char

    return ""


def twoSum(data, target):
    cache = {}
    for i in range(len(data)):
        diff = target - data[i]
        if diff in cache:
            return [cache[diff], i]
        cache[data[i]] = i
    return []


def shell_sort(arr):
    n = len(arr)
    gap = n // 2
    while gap > 0:
        for current_position in range(gap, n):
            m_gap = current_position
            while m_gap >= gap and arr[m_gap] < arr[m_gap - gap]:
                arr[m_gap], arr[m_gap - gap] = arr[m_gap - gap], arr[m_gap]
                m_gap -= gap
        gap = gap // 2
    return arr


def anagram(arr):
  hash = {}
  for i in arr:
    hash_key = ''.join(sorted(i))
    if hash_key not in hash:
        hash[hash_key] = []
    hash[hash_key].append(i)
  return list(hash.values())
```
