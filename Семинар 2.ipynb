```python
class Node(object):
  def __init__(self, d):
    self.data = d
    self.next = None

class LinkedList():
  def __init__(self):
    self.head = None

  def append_front(self, d):
    new_node = Node(d)
    if self.head is None:
      self.head = new_node
      return
    new_node.next = self.head
    self.head = new_node

  def append_back(self, data):
    new_node = Node(data)
    if self.head is None:
      self.head = new_node
      return
    cur_node = self.head
    while cur_node.next is not None:
      cur_node = cur_node.next
    cur_node.next = new_node

  def insert(self, after, n):
    search =  self.head
    while search is not None:
      if search.data == after:
        break
      search = search.next
      if search is not None:
        node = Node()
        node.data = n
        node.next = search.next
        search.next = node

  def get_head(self):
    return self.head

  def reverseLinkedList(self):
    prev = None
    current = self.head
    while current is not None:
      next = current.next
      current.next = prev
      prev = current
      current = next
    self.head = prev
    return self.head


  def midlelist(self):
    fast = slow = self.head
    while (fast is not None) and (fast.next is not None):
        slow = slow.next
        fast = fast.next.next

    print(f"Средний элемент: {slow.data}")
    return slow

  def display(self, max_nodes=20):
        """Вывод списка"""
        if self.head is None:
            print("Список пуст")
            return

        nodes = []
        current = self.head
        count = 0

        while current is not None and count < max_nodes:
            nodes.append(str(current.data))
            current = current.next
            count += 1

        if count == max_nodes:
            nodes.append("...")

        print(" -> ".join(nodes))





def has_cycle(head):
    if head is None:
        return False

    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

        if slow == fast:
            return True

    return False


def delete_node_from_list(head, val):
    while head is not None and head.data == val:
        head = head.next
    if head is None:
        return None
    current = head
    while current.next is not None:
        if current.next.data == val:
            current.next = current.next.next
        else:
            current = current.next

    return head

def display_list(head):
    nodes = []
    current = head
    while current is not None:
        nodes.append(str(current.data))
        current = current.next
    print(" -> ".join(nodes) if nodes else "Пустой список")


def isSubsequence(a, b):
    q = Queue()

    # Заполняем очередь символами из строки a
    for el in a:
        q.push(el)

    # Проходим по строке b
    for el in b:
        if not q.is_empty() and q.peek() == el:
            q.pop()

    return q.get_size() == 0

def isSubsequence_uk(a, b):
  i,j = 0,0
  while i < len(a) and j < len(b):
    if a[i] == b[j]:
      i+=1
    j+=1
  return i == len(a)

def isPalindrom_uk(a):
  i, j = 0, len(a)-1

  while i < j:
    if a[i] != a[j]:
      return False
    else:
      i += 1
      j -= 1
  return True



class Queue:
    def __init__(self):
        self.items = []

    def push(self, item):
        """Добавление элемента в очередь"""
        self.items.append(item)

    def pop(self):
        """Удаление и возврат первого элемента очереди"""
        if self.is_empty():
            return None
        return self.items.pop(0)

    def popright(self):
        """Удаление и возврат первого элемента очереди"""
        if self.is_empty():
            return None
        return self.items.pop(-1)

    def peek(self):
        """Просмотр первого элемента без удаления"""
        if self.is_empty():
            return None
        return self.items[0]

    def get_size(self):
        """Получение размера очереди"""
        return len(self.items)

    def is_empty(self):
        """Проверка на пустоту"""
        return len(self.items) == 0


def isPalindrome(word):
    if not word:
        return True

    # Сравниваем символы с обоих концов очереди
    while word.get_size() > 1:
        if word.pop() != word.popright():
            return False

    return True

def merge_two_sorted_lists(l1, l2):

    node = Node(0)
    n = node

    # Проходим по обоим спискам
    while l1 and l2:
        if l1.data <= l2.data:
            n.next = l1
            l1 = l1.next
        else:
            n.next = l2
            l2 = l2.next
        n = n.next

    # Добавляем оставшиеся элементы из того списка, который не закончился
    n.next = l1 if l1 else l2

    return display_list(node.next)
```
