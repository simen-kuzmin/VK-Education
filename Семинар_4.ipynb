```python
# ----- К задаче №1 ------------

def brute_force_search(text, pattern):
    """
    Метод грубой силы (прямого перебора)
    Сложность: O(n*m)
    """
    n = len(text)
    m = len(pattern)
    if m == 0:
        return 0
    if m > n:
        return -1
    for i in range(n - m + 1):
        j = 0
        while j < m and text[i + j] == pattern[j]:
            j += 1
        if j == m:
            return i
    return -1


# ----- К задаче №2 ------------


def boyer_moore_horspool(text, pattern):
    """
    Алгоритм Бойера-Мура-Хорспула
    Сложность: в лучшем случае O(n/m), в худшем O(n*m)
    """
    len_text = len(text)
    len_pattern = len(pattern)
    if len_pattern == 0:
        return 0 if len_text == 0 else -1
    if len_pattern > len_text:
        return -1
    offset = {}
    for i in range(len_pattern - 1):
        offset[pattern[i]] = len_pattern - i - 1
    default_offset = len_pattern
    i = 0
    while i <= len_text - len_pattern:
        j = len_pattern - 1
        while j >= 0 and pattern[j] == text[i + j]:
            j -= 1
        if j < 0:
            return i
        current_char = text[i + len_pattern - 1]
        shift = offset.get(current_char, default_offset)
        i += shift

    return -1

# ----- К задаче №3 ------------

def rabin_karp_hash(s, q=2147483647):
    """
    Вычисление полиномиального хеша для строки
    """
    x = 31
    result = ord(s[0])
    for i in range(len(s) - 1):
        result = (result * x + ord(s[i + 1])) % q
    return result


def rabin_karp_search(text, pattern):
    """
    Алгоритм Рабина-Карпа
    Сложность: в среднем O(n+m), в худшем O(n*m)
    """
    x = 31
    q = 2147483647
    n = len(text)
    m = len(pattern)
    if m == 0:
        return 0
    if m > n:
        return -1
    pattern_hash = rabin_karp_hash(pattern, q)
    x_power = pow(x, m - 1, q)
    current_hash = rabin_karp_hash(text[:m], q)
    for i in range(n - m + 1):
        if pattern_hash == current_hash:
            if pattern == text[i:i + m]:
                return i
        if i < n - m:
            current_hash = ((current_hash - ord(text[i]) * x_power) * x + ord(text[i + m])) % q
            if current_hash < 0:
                current_hash += q

    return -1


# ----- К задаче №4 ------------

def prefix_function(pattern):
    """
    Префикс-функция
    Возвращает массив длин наибольших собственных префиксов-суффиксов
    """
    n = len(pattern)
    pi = [0] * n
    for i in range(1, n):
        j = pi[i - 1]
        while j > 0 and pattern[j] != pattern[i]:
            j = pi[j - 1]
        if pattern[i] == pattern[j]:
            j += 1
        pi[i] = j

    return pi

# ----- К задаче №5 ------------

def kmp_search(text, pattern):
    """
    Алгоритм Кнута-Морриса-Пратта
    Сложность: O(n+m)
    """
    n = len(text)
    m = len(pattern)
    if m == 0:
        return 0
    if m > n:
        return -1
    pi = prefix_function(pattern)
    j = 0
    for i in range(n):
        while j > 0 and text[i] != pattern[j]:
            j = pi[j - 1]
        if text[i] == pattern[j]:
            j += 1
        if j == m:
            return i - m + 1

    return -1

# ----- К задаче №6 ------------

def z_function_trivial(s):
    """
    Z-функция (тривиальная реализация)
    Сложность: O(n^2)
    """
    n = len(s)
    z = [0] * n
    for i in range(1, n):
        while i + z[i] < n and s[z[i]] == s[i + z[i]]:
            z[i] += 1

    return z


def z_function(s):
    """
    Z-функция (оптимизированная реализация)
    Сложность: O(n)
    """
    n = len(s)
    z = [0] * n
    l, r = 0, 0  # границы самого правого z-блока
    for i in range(1, n):
        if i <= r:
            z[i] = min(r - i + 1, z[i - l])
        while i + z[i] < n and s[z[i]] == s[i + z[i]]:
            z[i] += 1
        if i + z[i] - 1 > r:
            l = i
            r = i + z[i] - 1

    return z


def z_search(text, pattern):
    """
    Поиск подстроки с помощью Z-функции
    """
    if not pattern:
        return 0
    s = pattern + '$' + text
    z = z_function(s)
    m = len(pattern)
    for i in range(m + 1, len(s)):
        if z[i] == m:
            return i - m - 1

    return -1
```
