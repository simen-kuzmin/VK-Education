```python
from collections import deque
import heapq
from typing import List
```

```python
class TreeNode:
  def __init__(self, val=0, left=None, right=None):
          self.val = val
          self.left = left
          self.right = right
          self.balanceFactor = None

# ----- К задаче №1 --------

def isMaxHeap(arr):
  n = len(arr);
  if n <= 1:
    return True
  queue = []
  queue.append(0)
  while (len(queue) > 0):
    i = queue.pop()
    left = 2 * i + 1
    right = 2 * i + 2
    if (left < n):
      if (arr[i] < arr[left]):
        return False
      queue.append(left)
    if (right < n):
      if (arr[i] < arr[right]):
        return False
      queue.append(right)
  return True



# ----- К задаче №2 ---------

def isCompleteTree(root):
    if not root:
        return True
    queue = deque([root])
    shouldBeLeaf = False
    while queue:
        node = queue.popleft()
        if not node:
            shouldBeLeaf = True
        else:
            if shouldBeLeaf:
                return False
            queue.append(node.left)
            queue.append(node.right)
    return True


# ----- К задаче №3 --------

def mergeKSortedArrays(sortedArrays: List[List[int]]) -> List[int]:
    if not sortedArrays:
        return []
    mergedArray = []
    minHeap = []
    for i, currentArray in enumerate(sortedArrays):
        if currentArray:
            heapq.heappush(minHeap, (currentArray[0], i, 0))
    while minHeap:
        value, arrayIdx, elementIdx = heapq.heappop(minHeap)
        mergedArray.append(value)
        if elementIdx + 1 < len(sortedArrays[arrayIdx]):
            nextElement = sortedArrays[arrayIdx][elementIdx + 1]
            heapq.heappush(minHeap, (nextElement, arrayIdx, elementIdx + 1))

    return mergedArray


# ----- К задаче №4 ----------


def inorderMin(node, k, counter):
    if node is None:
        return None
    left_result = inorderMin(node.left, k, counter)
    if left_result is not None:
        return left_result
    counter[0] += 1
    if counter[0] == k:
        return node.val
    return inorderMin(node.right, k, counter)



def buildTree(arr, i):
    if (i >= len(arr)):
      return None
    root = TreeNode(arr[i])
    root.left = buildTree(arr, 2 * i + 1)
    root.right = buildTree(arr, 2 * i + 2)
    return root



# ----- К задаче №5 ---------

def calculateHeightsAndBalance (node):
  if not node:
    return 0
  leftHeight = calculateHeightsAndBalance (node.left)
  rightHeight = calculateHeightsAndBalance (node.right)
  node.balanceFactor = leftHeight - rightHeight
  return 1 + max(leftHeight, rightHeight)


# ----- К задаче №6 ---------

def mirrorTreeIterative(node):
    if node is None:
        return None

    # Очередь для обхода в ширину (BFS)
    queue = deque([node])

    while queue:
        current = queue.popleft()

        # Меняем местами левый и правый потомки
        temp = current.left
        current.left = current.right
        current.right = temp

        # Добавляем в очередь дочерние узлы
        if current.left:
            queue.append(current.left)
        if current.right:
            queue.append(current.right)

    return node
```

## Задача 6

```python
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

print("До зеркального отражения:")
print(f"     {root.val}")
print(f"   /   \\")
print(f"  {root.left.val}     {root.right.val}")
print(f" / \\")
print(f"{root.left.left.val}   {root.left.right.val}")

mirrorTreeIterative(root)

print("\nПосле зеркального отражения:")
print(f"     {root.val}")
print(f"   /   \\")
print(f"  {root.left.val}     {root.right.val}")
print(f"       / \\")
print(f"      {root.right.left.val}   {root.right.right.val}")

# Проверяем результаты
assert root.val == 1
assert root.left.val == 3  # Было 2, стало 3
assert root.right.val == 2  # Было 3, стало 2
assert root.right.left.val == 5  # Было 4, стало 5
assert root.right.right.val == 4  # Было 5, стало 4
```
