```python
# ----- К задаче №1 ------------

def b_sequences(n):
  if n == 1:
    return 2
  if n == 2:
    return 3
  return b_sequences(n - 1) + b_sequences(n - 2)


# ----- К задаче №2 ------------

def count_sequences(n):
  if (n == 0):
    return 1
  if (n == 1):
    return 2
  if (n == 2):
    return 4
  dp = [1, 2, 4]
  for i in range(3, n+1):
    dp.append(dp[i - 1] + dp[i - 2] + dp[i - 3])
  return dp[n]

# ----- К задаче №3 ------------

def findLIS(nums):
  if (len(nums) == 0):
    return 0
  if (len(nums) == 1):
    return 1
  dp = [1] * len(nums)
  for i in range(1, len(nums)):
    if (nums[i - 1] < nums[i]):
      dp[i] = dp[i - 1] + 1
  return max(dp)

# ----- К задаче №4 ------------

def factorial(num):
    if num == 0 or num == 1:
        return 1
    result = 1
    for i in range(2, num + 1):
        result *= i
    return result

def pascal_triangle_combinatorial(n):
    triangle = []

    for i in range(n):
        row = []
        for j in range(i + 1):
            # C(i, j) = i! / (j! * (i-j)!)
            value = factorial(i) // (factorial(j) * factorial(i - j))
            row.append(value)
        triangle.append(row)

    return triangle

# ----- К задаче №5 ------------

def max_profit(prices):
    if not prices or len(prices) < 2:
        return 0

    min_price = float('inf')
    max_profit = 0

    for price in prices:
        if price < min_price:
            min_price = price
        potential_profit = price - min_price
        if potential_profit > max_profit:
            max_profit = potential_profit

    return max_profit

# ----- К задаче №6 ------------

def coin_change(coins, amount):
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0
    for current_amount in range(1, amount + 1):
        for coin in coins:
            if coin <= current_amount:
                dp[current_amount] = min(dp[current_amount],
                                       dp[current_amount - coin] + 1)
    return dp[amount] if dp[amount] != float('inf') else -1

# ----- К задаче №7 ------------

def longest_palindrome(s):
    if len(s) < 2:
        return s

    result = ""

    for i in range(len(s)):
        l, r = i, i
        while l >= 0 and r < len(s) and s[l] == s[r]:
            if (r - l + 1) > len(result):
                result = s[l:r+1]
            l -= 1
            r += 1
        l, r = i, i+1
        while l >= 0 and r < len(s) and s[l] == s[r]:
            if (r - l + 1) > len(result):
                result = s[l:r+1]
            l -= 1
            r += 1

    return result
```
