```python
def TwoSum(array , target):
  l = len(array)
  i, j = 0, l-1

  while i < j:
    sum = array[i] + array[j]

    if sum > target:
      j -=1
    elif sum < target:
      i += 1
    else:
      return i, j


def ReverseArray(array, left = None, right = None):
    l = len(array)
    if left == None:
      i, j = 0, l-1
    else:
      i, j = left, right

    while i < j:
      a = array[i]
      array[i] = array[j]
      array[j] = a

      i += 1
      j -= 1
    return array

def PartReverse(array, k):
  l = len(array)

  ReverseArray(array, 0, l-1)
  ReverseArray(array, 0, k % l - 1 )
  ReverseArray(array, k % l , l-1)

  return array

def merge(arr1, arr2):
  i = len(arr1) - len(arr2)-1
  k = len(arr1) - 1
  j = len(arr2) - 1

  while j >= 0:
    if  (i >= 0) and (arr1[i] > arr2[j]):
      arr1[k] = arr1[i]
      i -= 1
    else:
      arr1[k] = arr2[j]
      j -= 1
    k -= 1

  return arr1


def sort_binary(array):
    l = len(array)
    i, j = 0, l-1

    while i < j:
      if array[i] == 0:
        i += 1
      elif array[j] == 1:
        j -= 1
      else:
        array[i], array[j] = array[j], array[i]
        j -= 1
        i += 1
    return array


def sortColor(arr):
  i, j = 0, 0
  k = len(arr) - 1

  while j < k:
    if arr[j] == 0:
      arr[i], arr[j] =  arr[j], arr[i]
      i += 1
      j += 1
    elif arr[j] == 1:
      j += 1
    elif arr[j] == 2:
      arr[j], arr[k] =  arr[k], arr[j]
      k -= 1
  return arr

def evenFirst(arr):
  j = 0
  for i in range(len(arr)):
    if arr[i] % 2 == 0:
      arr[i], arr[j] = arr[j], arr[i]
      j += 1

  return arr


def sort_Unbinary(arr):
    l = len(arr)
    j = 0

    for i in range(l):
      if arr[i] != 0:
        arr[i], arr[j] = arr[j], arr[i]
        j += 1


    return arr
```
